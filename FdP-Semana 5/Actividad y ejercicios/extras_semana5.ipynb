# Extra 1: Sistema de calificaciones con tuplas
Enunciado: Crea una tupla calificaciones = (7.5, 9.0, 8.0, 6.5, 10.0). Imprime la tercera calificación, captura dos calificaciones nuevas con input() y anéxalas en una nueva tupla. Convierte la nueva tupla en lista, ordénala de mayor a menor y muestra la suma de todos los elementos usando una función que retorne el valor Entrada:

Nueva calificación 1: 8.5
Nueva calificación 2: 9.5

Salida:

Tercera calificación: 8.0 

Nueva tupla: (7.5, 9.0, 8.0, 6.5, 10.0, 8.5, 9.5)

Lista ordenada (mayor a menor): [10.0, 9.5, 9.0, 8.5, 8.0, 7.5, 6.5]

Suma total: 59.0

```python
Calificaciones = (7.5, 9.0, 8.0, 6.5, 10.0)

# Se imprme la tercera calificación recordando que 0 es 1
print(f"Tercera calificación:", Calificaciones[2])



Calificacion1 = float(input("Coloca la quinta calificación:"))
Calificacion2 = float(input("Coloca la sexta calificación:"))

# Nueva tupla con nuevas calificaciones
Nuevas_calificaciones = Calificaciones + (Calificacion1, Calificacion2)

print(f"Nueva tupla:",Nuevas_calificaciones)


# Calificaciones ordenadas
Orden = list(Nuevas_calificaciones)
Orden.sort(reverse=True)
print(f"Lista ordenada (mayor a menor):",Orden)

# Suma de calificaciones
Suma_Calificaciones = sum(Orden)

print(f"Suma total:",Suma_Calificaciones)






```

# Extra 2: Agenda de contactos con búsqueda
Enunciado: Crea un diccionario agenda con tres contactos iniciales ("Ana": "555-0101", "Luis": "555-0102", "Mía": "555-0103"). Captura un nuevo contacto (nombre y teléfono) con input() y agrégalo al diccionario. Luego imprime todos los nombres con keys() en una sola línea y, con una función que retorne el valor, busca y muestra el teléfono del nuevo contacto. Entrada:

Nombre del nuevo contacto: Pedro
Teléfono del nuevo contacto: 555-0140
Nombre a buscar: Pedro

Salida:

Contactos registrados: Ana, Luis, Mía, Pedro
El teléfono de Pedro es: 555-0140

```python

# Diccionario de contactos inicial
contactos = {
    "Ana": "555-0101",
    "Luis": "555-0102",
    "Mía": "555-0103"
}

# Ingresa el nombre y número del nuevo contacto

Nombre = input("Nombre del nuevo contacto:")
Numero = input("Ingresa el Número del nuevo contacto:")

# Se agrega en nuevo contacto al diccionario y el el Nuevo contacto y su número a buscar
contactos[Nombre] = Numero

# Busca el numero de telefono
def obtener_telefono(nombre, diccionario_contactos):
    return diccionario_contactos.get(nombre, "Contacto no encontrado")

# Aqui se imprime el Nuevo Diccionario solo con nombres gracias a ".keys".
print(f"Contactos registrados: {', '.join(contactos.keys())}")

telefono_encontrado = obtener_telefono(Nombre, contactos)
print(f"El teléfono de {Nombre} es: {telefono_encontrado}")


```

# Extra 3: Calculadora segura con manejo de excepciones
Enunciado: Escribe un programa que pida dos números enteros. Implementa un bloque try-except que capture ValueError (caracteres no numéricos o vacíos) con un mensaje controlado y un bloque específico except ZeroDivisionError que muestre un mensaje amigable si el segundo número es cero. En caso exitoso, muestra la división del primero entre el segundo. Entrada:

Primer número entero: 20
Segundo número entero: 4

- Salida:

20 dividido entre 4 es: 5.0

- Caso de error (divisor cero):

Primer número entero: 10
Segundo número entero: 0

- Salida:

Error: No es posible dividir entre cero. Ingresa un divisor distinto de 0.

```python
try:
    Numero1 = int(input("Ingresa un número: "))
    Numero2 = int(input("Ingresa otro número:"))
    print(f"{Numero1} dividido entre {Numero2} es: ", Numero1 / Numero2)

# ZeroDivisionError: se intenta dividir un número entre cero.
except ZeroDivisionError:
    print("Error: No es posible dividir entre cero. Ingresa un divisor distinto de 0.")
```

# Desafío Extra 2: Contador de votos con diccionario
Crea un diccionario votos = {"Rojo": 0, "Azul": 0, "Verde": 0}. Pide al usuario el color de su voto cinco veces, suma uno al valor correspondiente en cada ocasión y, al final, imprime con items() el color ganador junto con su total.

```python
Votos = {
    "Rojo": 0,
    "Azul": 0,
    "Verde": 0
}

for c in range(5):
    voto = input("---OPCIONES---\nRojo\nAzul\nVerde\nIngresa el color a votar: ")

    if voto == "Rojo":
        Votos["Rojo"] += 1

    elif voto == "Azul":
        Votos["Azul"] += 1

    elif voto == "Verde":
        Votos["Verde"] += 1

    else:
        print("Opción no válida. Intenta nuevamente.")


C_Ganador = "Rojo"
total_votos = Votos["Rojo"]

if Votos["Azul"] > total_votos:
    C_Ganador = "Azul"
    total_votos = Votos["Azul"]

if Votos["Verde"] > total_votos:
    C_Ganador = "Verde"
    total_votos = Votos["Verde"]

print(f"El color {C_Ganador} gana con {total_votos} votos.")
```
