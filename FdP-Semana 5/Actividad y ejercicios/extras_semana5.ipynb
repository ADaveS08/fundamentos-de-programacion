# Extra 1: Sistema de calificaciones con tuplas
Enunciado: Crea una tupla calificaciones = (7.5, 9.0, 8.0, 6.5, 10.0). Imprime la tercera calificación, captura dos calificaciones nuevas con input() y anéxalas en una nueva tupla. Convierte la nueva tupla en lista, ordénala de mayor a menor y muestra la suma de todos los elementos usando una función que retorne el valor Entrada:

Nueva calificación 1: 8.5
Nueva calificación 2: 9.5

Salida:

Tercera calificación: 8.0 

Nueva tupla: (7.5, 9.0, 8.0, 6.5, 10.0, 8.5, 9.5)

Lista ordenada (mayor a menor): [10.0, 9.5, 9.0, 8.5, 8.0, 7.5, 6.5]

Suma total: 59.0

```python
Calificaciones = (7.5, 9.0, 8.0, 6.5, 10.0)

#Tercera calificación
print(f"Tercera calificación:", Calificaciones[2])



Calificacion1 = float(input("Coloca la quinta calificación:"))
Calificacion2 = float(input("Coloca la sexta calificación:"))

# Nueva tupla con nuevas calificaciones
Nuevas_calificaciones = Calificaciones + (Calificacion1, Calificacion2)

print(f"Nueva tupla:",Nuevas_calificaciones)


# Calificaciones ordenadas
Orden = list(Nuevas_calificaciones)
Orden.sort(reverse=True)
print(f"Lista ordenada (mayor a menor):",Orden)

# Suma de calificaciones
Suma_Calificaciones = sum(Orden)

print(f"Suma total:",Suma_Calificaciones)






```
