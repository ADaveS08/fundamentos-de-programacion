```python
# ACTIVIDAD 4 - MENU MODULAR


# ==========================================================
# OPCION 1 - TUPLAS
# ==========================================================

def trabajar_tuplas():

    # Tupla inicial
    numeros = (7.5, 9.0, 8.0, 6.5, 10.0)

    # Imprimir el tercer elemento
    print("\nTercer elemento:", numeros[2])

    # Capturar dos números adicionales
    Numero1 = float(input("Ingresa el sexto número: "))
    Numero2 = float(input("Ingresa el séptimo número: "))

    # Crear una nueva tupla
    nueva_tupla = numeros + (Numero1, Numero2)

    print("Nueva tupla:", nueva_tupla)

    # Convertir la tupla en lista
    lista_numeros = list(nueva_tupla)

    # Ordenar de mayor a menor
    lista_numeros.sort(reverse=True)

    print("Lista ordenada (mayor a menor):", lista_numeros)

    # Función para sumar los números
    def sumar_numeros(tupla):
        return sum(tupla)

    # Mostrar la suma
    suma = sumar_numeros(nueva_tupla)

    print("Suma total:", suma)


# ==========================================================
# OPCION 2 - DICCIONARIOS
# ==========================================================

def trabajar_diccionarios():

    # Diccionario de contactos inicial
    contactos = {
        "Ana": "555-0101",
        "Luis": "555-0102",
        "Mía": "555-0103"
    }

    # Capturar nuevo contacto
    Nombre = input("\nNombre del nuevo contacto: ")
    Numero = input("Teléfono del nuevo contacto: ")

    # Agregar el nuevo contacto
    contactos[Nombre] = Numero

    # Mostrar solamente los nombres
    print("\nContactos registrados:")

    for nombre in contactos.keys():
        print(nombre, end=", ")

    print()

    # Función para buscar teléfono
    def obtener_telefono(diccionario_contactos, nombre):
        return diccionario_contactos.get(nombre, "Contacto no encontrado")

    # Buscar el nuevo contacto
    Nombre_buscar = input("Nombre a buscar: ")

    telefono_encontrado = obtener_telefono(contactos, Nombre_buscar)

    print(f"El teléfono de {Nombre_buscar} es: {telefono_encontrado}")


# ==========================================================
# OPCION 3 - EXCEPCIONES
# ==========================================================

def trabajar_excepciones():

    try:

        # Solicitar dos números enteros
        Num1 = int(input("\nPrimer número entero: "))
        Num2 = int(input("Segundo número entero: "))

        # Realizar la división
        Resultado = Num1 / Num2

        print(f"{Num1} dividido entre {Num2} es: {Resultado}")

    except ValueError:

        print("Error: Debes ingresar números enteros. No se permite texto ni valores vacíos.")

    except ZeroDivisionError:

        print("Error: No es posible dividir entre cero. Ingresa un divisor distinto de 0.")


# ==========================================================
# OPCION 4 - STRINGS
# ==========================================================

def trabajar_strings():

    # Mensaje inicial
    mensaje = "Python es un lenguaje poderoso"

    # Longitud del mensaje
    print("\nLongitud del mensaje:", len(mensaje))

    # Convertir a mayúsculas
    print("En mayúsculas:", mensaje.upper())

    # Reemplazar una palabra
    texto_reemplazado = mensaje.replace("Python", "programación")

    print("Texto reemplazado:", texto_reemplazado)

    # Función para contar palabras
    def contar_palabras(texto):
        return len(texto.split())

    # Mostrar cantidad de palabras
    total_palabras = contar_palabras(mensaje)

    print("Palabras totales:", total_palabras)


# ==========================================================
# MENU PRINCIPAL
# ==========================================================

while True:

    print("\n===================================")
    print("       MENU PRINCIPAL")
    print("===================================")
    print("(1) Trabajar con tuplas")
    print("(2) Trabajar con diccionarios")
    print("(3) Manejo de excepciones")
    print("(4) Analizar strings")
    print("(5) Finalizar el programa")
    print("===================================")

    try:
        Opcion = int(input("Ingresa la opción numérica: "))

        if Opcion == 1:

            trabajar_tuplas()

        elif Opcion == 2:

            trabajar_diccionarios()

        elif Opcion == 3:

            trabajar_excepciones()

        elif Opcion == 4:

            trabajar_strings()

        elif Opcion == 5:

            print("\nPrograma finalizado.")
            break

        else:

            print("\nError: Selecciona una opción del 1 al 5.")

    except ValueError:

        print("\nError: Debes ingresar un número del 1 al 5.")
```
