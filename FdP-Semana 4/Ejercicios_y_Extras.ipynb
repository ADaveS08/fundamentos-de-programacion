# Matriz Ejemplo:

```python
matriz = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]
# Recorre lineas
for fila in matriz: 
    for elemento in fila: #Este recorre columnas
        print(elemento, end=" ")
    print()



    
```

```python
filas = 3
columnas = 3


matriz = []
for f in range(filas):
    fila_actual = []
    for c in range(columnas):
        valor = int(input(f"Valor para [{f}][{c}]"))
        fila_actual.append(valor)
    matriz.append(fila_actual)

print("Impresion de matriz")
for f in range(filas):
    for c in range(columnas):
        print(matriz[f][c], end=" ")
    print()
```

# Codigo que imprima una matriz de 0 con una interseccion en linea diagonal de 1.

```python
dimension = int(input("ingresa cuanto quieres que mida tu matriz:"))

matriz = []
for f in range(dimension):
    fila_actual = []
    for c in range(dimension):
        valor = 0
        if f == c:
            valor = 1
        fila_actual.append(valor)
    matriz.append(fila_actual)

print("Impresion de matriz")
for f in range(dimension):
    for c in range(dimension):
        print(matriz[f][c], end=" ")
    print()

```

#  Programa que pida diez numeros en un rango de -10 a 10 y que imprima la cantidad de asteriscos de acuerdo al numero capturado


```python
lista = []
contador = 1

while contador <= 10:
    Num = int(input("Ingresa 10 números en un rango del -10 al 10: "))

    if Num >= -10 and Num <= 10:
        contador += 1
        lista.append(Num)
    else:
        print("Rango incorrecto, ingresa correctamente entre -10 y 10:")

print("\n", lista)

for i in lista:
    if i < 0:
        print(" " * (10 - abs(i)) + "*" * abs(i) + "|")
    elif i > 0:
        print(" " * 10 + "|" + "*" * i)
    else:
        print(" " * 10 + "|")
```
