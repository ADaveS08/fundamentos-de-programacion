### Desafío Extra 1: Convertidor de Temperatura

```python
#Formula Fahrenheit
C = float(input("Ingresa la temperatura en grados Celsius: "))
F = (C * 9/5) + 32
K = C + 273.15

print(f"Temperatura en Celsius: {C}grados Celsius")
print(f"Temperatura en Fahrenheit: {F}grados Fahrnheit")
print(f"Temperatura en Kelvin: {K}Kelvin")
```

# ACTIVIDAD EVALUABLE 1 — CALCULADORA DE TIEMPO DIGITAL

```python
nombre_de_usuario = input("Ingresa tu nombre de usuario:")

redes_sociales = float(input("(En Horas)Ingresa cuanto tiempo le dedicas a redes sociales:"))
WhatsApp = float(input("(En Horas)Ingresa cuanto tiempo le dedicas a instagram:"))
TikTok = float(input("(En Horas)ingresa cuanto tiempo le dedicas al TikTok:"))
Canvas = float(input("(En Horas) Ingresa cuanto timpo le dedicas a esta app:"))
Facebook = float(input("(En Horas) Ingresa cuanto tiempo le dedics al facebook:"))

Tiempo_total = (redes_sociales + WhatsApp + TikTok + Canvas + Facebook)

print(nombre_de_usuario)
print(f"pasas un total de {Tiempo_total} horas")
porcentaje = ( Tiempo_total / 24) * 100
print(f"ocupas el {porcentaje} % del dia ")
```
